```python
import os

# Base directory of your dataset
base_dir = r"D:\PHD Project\Balanced"

# Function to print folder structure
def print_folder_structure(base_dir):
    for root, dirs, files in os.walk(base_dir):
        # Calculate the indentation level
        level = root.replace(base_dir, '').count(os.sep)
        indent = '    ' * level
        print(f"{indent}📂 {os.path.basename(root)}/")
        for f in files[:5]:  # show only first 5 files per folder
            print(f"{indent}    🖼️ {f}")
        if len(files) > 5:
            print(f"{indent}    ... {len(files)-5} more files")

# Print the structure
print_folder_structure(base_dir)
```
